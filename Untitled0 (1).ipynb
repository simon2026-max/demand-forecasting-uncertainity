```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv", parse_dates=["date"])
print(df.shape)
print(df.head())
print(df.isna().sum())
print(df["date"].min(), "to", df["date"].max())
print("stores:", df["store"].nunique(), "items:", df["item"].nunique())
one = df[(df["store"] == 1) & (df["item"] == 1)]
plt.figure(figsize=(12, 4))
plt.plot(one["date"], one["sales"])
plt.title("Store 1, Item 1: daily sales")
plt.show()
```

```python
from pandas.tseries.holiday import USFederalHolidayCalendar

df = df.sort_values(["store", "item", "date"]).reset_index(drop=True)

# calendar features
df["dayofweek"] = df["date"].dt.dayofweek
df["month"] = df["date"].dt.month
df["dayofyear"] = df["date"].dt.dayofyear
df["weekofyear"] = df["date"].dt.isocalendar().week.astype(int)
df["year"] = df["date"].dt.year
df["is_weekend"] = (df["dayofweek"] >= 5).astype(int)

# holidays
hol = USFederalHolidayCalendar().holidays(start=df["date"].min(), end=df["date"].max())
df["is_holiday"] = df["date"].isin(hol).astype(int)

# lags and rolling stats
g = df.groupby(["store", "item"])["sales"]
for lag in [7, 14, 28, 365]:
    df[f"lag_{lag}"] = g.shift(lag)
for w in [7, 28]:
    df[f"roll_mean_{w}"] = g.transform(lambda s: s.shift(7).rolling(w).mean())
    df[f"roll_std_{w}"] = g.transform(lambda s: s.shift(7).rolling(w).std())

df = df.dropna().reset_index(drop=True)
print(df.shape)
print(df.head())
```

```python
import numpy as np
import lightgbm as lgb

features = ["store", "item", "dayofweek", "month", "dayofyear", "weekofyear",
            "year", "is_weekend", "is_holiday",
            "lag_7", "lag_14", "lag_28", "lag_365",
            "roll_mean_7", "roll_mean_28", "roll_std_7", "roll_std_28"]

# time-based split: train on the past, test on the last 3 months of 2017
train = df[df["date"] < "2017-10-01"]
test = df[df["date"] >= "2017-10-01"]

models = {}
preds = {}
for q in [0.1, 0.5, 0.9]:
    m = lgb.LGBMRegressor(objective="quantile", alpha=q,
                          n_estimators=300, learning_rate=0.1,
                          num_leaves=63, verbose=-1)
    m.fit(train[features], train["sales"])
    models[q] = m
    preds[q] = m.predict(test[features])
    print("done", q)

# keep the three forecasts in order: low <= middle <= high
p = np.sort(np.column_stack([preds[0.1], preds[0.5], preds[0.9]]), axis=1)
test = test.copy()
test["p10"], test["p50"], test["p90"] = p[:, 0], p[:, 1], p[:, 2]

# 1) point accuracy (middle forecast)
mae = np.mean(np.abs(test["sales"] - test["p50"]))
rmse = np.sqrt(np.mean((test["sales"] - test["p50"]) ** 2))
print("MAE:", round(mae, 2), " RMSE:", round(rmse, 2))

# 2) calibration: should be close to 80%
inside = ((test["sales"] >= test["p10"]) & (test["sales"] <= test["p90"])).mean()
print("Actuals inside the 10-90 interval:", round(inside * 100, 1), "%")
```

```python
# Assumed costs per unit (made-up numbers, you can change them)
cost_stockout = 5   # lost profit when we run out
cost_overstock = 1  # holding cost for unsold stock

def total_cost(stock, actual):
    short = np.maximum(actual - stock, 0)   # units we couldn't sell
    extra = np.maximum(stock - actual, 0)   # units left over
    return (cost_stockout * short + cost_overstock * extra).sum()

actual = test["sales"].values
for name in ["p10", "p50", "p90"]:
    stock = np.ceil(test[name].values)
    cost = total_cost(stock, actual)
    short_days = (actual > stock).mean() * 100
    print(f"Stock at {name}: total cost = {cost:,.0f}, stockout on {short_days:.1f}% of days")
```

```python
test["inside"] = (test["sales"] >= test["p10"]) & (test["sales"] <= test["p90"])
test["spike"] = test["sales"] > 1.3 * test["roll_mean_28"]

print("Coverage on normal days:", round(test[~test["spike"]]["inside"].mean() * 100, 1), "%")
print("Coverage on demand spikes:", round(test[test["spike"]]["inside"].mean() * 100, 1), "%")
print()
print("Coverage by holiday:")
print((test.groupby("is_holiday")["inside"].mean() * 100).round(1))
print()
print("Coverage by day of week (0 = Monday):")
print((test.groupby("dayofweek")["inside"].mean() * 100).round(1))
```

```python
s = test[(test["store"] == 1) & (test["item"] == 1)]
plt.figure(figsize=(12, 4))
plt.fill_between(s["date"], s["p10"], s["p90"], alpha=0.3, label="10-90% interval")
plt.plot(s["date"], s["p50"], label="Forecast (p50)")
plt.plot(s["date"], s["sales"], "k.", markersize=4, label="Actual")
plt.legend()
plt.title("Store 1, Item 1: forecast with prediction interval")
plt.show()
```
